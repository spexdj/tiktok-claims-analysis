Use `groupby()` to group the data by `author_ban_status`, then use `agg()` to get the count, mean, and median of each of the following columns:
* `video_view_count`
* `video_like_count`
* `video_share_count`

Remember, the argument for the `agg()` function is a dictionary whose keys are columns. The values for each column are a list of the calculations you want to perform.

```python

data.groupby("author_ban_status").agg({
    "video_view_count": ["count", "mean", "median"],
    "video_like_count": ["count", "mean", "median"],
    "video_share_count": ["count", "mean", "median"]
})
```

Las diferencias en las medianas son aún más pronunciadas: los videos de autores baneados registran una mediana de 448.201 visualizaciones frente a 8.616; 105.573 "me gusta" frente a 2.222; y 14.468 veces compartido frente a 437. Los videos de autores bajo revisión se sitúan, en estas métricas, en un punto intermedio entre los grupos de autores baneados y los activos.

Se trata de asociaciones descriptivas que podrían reflejar, en parte, las distintas proporciones de afirmaciones y opiniones presentes en cada grupo. Comparar la interacción por visualización —desglosada según el tipo de afirmación y el estatus del autor— permitirá distinguir la intensidad de la interacción de las diferencias en el alcance ante la audiencia.


Now, create three new columns to help better understand engagement rates:
* `likes_per_view`: represents the number of likes divided by the number of views for each video
* `comments_per_view`: represents the number of comments divided by the number of views for each video
* `shares_per_view`: represents the number of shares divided by the number of views for each video

```python
valid_views = data["video_view_count"].where(data["video_view_count"] > 0)

# Create a likes_per_view column
data["likes_per_view"] = data["video_like_count"] / valid_views

# Create a comments_per_view column
data["comments_per_view"] = data["video_comment_count"] / valid_views

# Create a shares_per_view column
data["shares_per_view"] = data["video_share_count"] / valid_views
```

Use `groupby()` to compile the information in each of the three newly created columns for each combination of categories of claim status and author ban status, then use `agg()` to calculate the count, the mean, and the median of each group.

```python

data.groupby(["claim_status", "author_ban_status"]).agg({
    "likes_per_view": ["count", "mean", "median"],
    "comments_per_view": ["count", "mean", "median"],
    "shares_per_view": ["count", "mean", "median"]
})
```

**Question:**

How does the data for claim videos and opinion videos compare or differ? Consider views, comments, likes, and shares.
En promedio, los videos que presentan afirmaciones reciben aproximadamente 101 veces las visualizaciones que los videos de opinión. Tras normalizar el nivel de interacción en función del número de visualizaciones, los videos de afirmaciones también presentan valores más altos —tanto en la media como en la mediana— de "me gusta", comentarios y veces compartido por cada visualización, independientemente del estado de bloqueo del autor.

En el caso de los videos de afirmaciones, la media de "me gusta" por visualización oscila aproximadamente entre 0,328 y 0,345, frente a un rango de 0,207 a 0,226 para los videos de opinión. La media de comentarios por visualización varía entre 0,00137 y 0,00139 para los videos de afirmaciones, frente a un rango de 0,00043 a 0,00054 para los de opinión. Por su parte, la media de veces compartido por visualización se sitúa entre 0,0655 y 0,0679 para los videos de afirmaciones, frente a un rango de 0,0405 a 0,0445 para los de opinión.

Las diferencias según el estado de bloqueo del autor dentro de cada categoría son menores, y los autores bloqueados no presentan sistemáticamente las tasas de interacción más elevadas. Esto sugiere que las cifras absolutas de interacción más altas observadas en los autores bloqueados deben interpretarse teniendo en cuenta el alcance de la audiencia y la composición de los videos de afirmaciones y de opinión. Estos hallazgos representan asociaciones descriptivas dentro de un conjunto de datos sintético y no establecen una relación causal.
